```python
from pathlib import Path
import json
import os
```

```python
SUBMITTED_FILE = Path("/home/vince/Documents/SmartFires/osm_fitness/Azira/get-urls/states_submitted_011526.txt")
URLS_FILE = Path("/home/vince/Documents/SmartFires/osm_fitness/Azira/get-urls/urls.txt")

result = []

with open(SUBMITTED_FILE, "r") as f:
    submitted_states = set(line.strip() for line in f)

with open(URLS_FILE, "r") as f:
    all_urls = f.readlines()

for l in all_urls:
    if any(state in l for state in submitted_states):
        result.append(l)
```

```python
with open("urls_to_download_012226.txt", "w") as f:
    f.writelines(result)
```
