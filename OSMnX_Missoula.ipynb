# Calculate Area per Category (Notice the 100% overlap on sports centers... remove?)

```python


# Make sure both are projected
gdf = gdf.to_crs(3857)
gdf_osm = gdf_osm.to_crs(3857)

# Total area of ParkServe and OSM polygons
total_parkserve_area = gdf.area.sum()

results = []

for cat, subset in gdf_osm.groupby("osm_category"):
    # intersection
    overlap = gpd.overlay(gdf, subset, how="intersection")
    overlap_area = overlap.area.sum()
    osm_area = subset.area.sum()
    
    results.append({
        "osm_category": cat,
        "overlap_km2": overlap_area / 1e6,
        "osm_area_km2": osm_area / 1e6,
        "parkserve_overlap_pct": 100 * overlap_area / total_parkserve_area,
        "osm_overlap_pct": 100 * overlap_area / osm_area if osm_area > 0 else 0,
    })

df_summary = pd.DataFrame(results).sort_values("overlap_km2", ascending=False)
print(df_summary)
```

## Remove large parks and replace with trailhead polygons (also just add polygons)

```python
import osmnx as ox
import geopandas as gpd
from shapely.geometry import box

def get_trailhead_polygons(
    place,
    gdf_parkserve,
    gdf_osm,
    area_threshold=200000,
    grid_size=150,
    crs_epsg=3857
):
    """
    Generates polygons for all trailheads (even outside parks),
    removes large parks that already contain trailheads from both datasets,
    and returns:
        (modified_parkserve, modified_osm, trailhead_polygons)
    """

    # --- Ensure CRSs are defined ---
    if gdf_parkserve.crs is None:
        gdf_parkserve = gdf_parkserve.set_crs(4326)
    if gdf_osm.crs is None:
        gdf_osm = gdf_osm.set_crs(4326)

    # --- Get trailheads from OSM ---
    tag_sets = [
        {"highway": "trailhead"}    ]

    dfs = []
    for tags in tag_sets:
        try:
            gdf_tmp = ox.features_from_place(place, tags)
            if not gdf_tmp.empty:
                dfs.append(gdf_tmp)
        except Exception as e:
            print(f"Could not fetch {tags}: {e}")

    if not dfs:
        print("No OSM features found — returning empty GeoDataFrame.")
        return gpd.GeoDataFrame(columns=["geometry"], crs="EPSG:4326")

    gdf_trailheads = gpd.GeoDataFrame(pd.concat(dfs, ignore_index=True))
    gdf_trailheads = gdf_trailheads[gdf_trailheads.geometry.notnull()].copy()


    if gdf_trailheads.empty:
        print("No trailheads found — returning original datasets.")
        return gdf_parkserve, gdf_osm, gpd.GeoDataFrame(columns=["geometry"], crs=gdf_parkserve.crs)

    # --- Reproject everything ---
    gdf_trailheads = gdf_trailheads.to_crs(crs_epsg)
    gdf_parkserve = gdf_parkserve.to_crs(crs_epsg)
    gdf_osm = gdf_osm.to_crs(crs_epsg)

    # --- Create small polygons around each trailhead ---
    boxes = []
    for pt in gdf_trailheads.geometry:
        if pt.is_empty:
            continue
        x, y = pt.x, pt.y
        half = grid_size / 2
        boxes.append(box(x - half, y - half, x + half, y + half))

    trailhead_polys = gpd.GeoDataFrame(geometry=boxes, crs=crs_epsg)

   # --- Remove large parks that are near trailheads ---
    # Identify large parks
    large_ps = gdf_parkserve[gdf_parkserve.geometry.area > area_threshold]
    large_osm = gdf_osm[gdf_osm.geometry.area > area_threshold]
    
    # Define a buffer distance (meters) around trailheads
    buffer_dist = 200  # adjust this as you like — e.g., 100m, 250m, etc.
    
    # Create a combined buffered area around all trailheads
    trailhead_buffer = gdf_trailheads.buffer(buffer_dist).unary_union
    
    # Find large parks whose geometries are within buffer distance
    ps_near_th = large_ps[large_ps.intersects(trailhead_buffer)]
    osm_near_th = large_osm[large_osm.intersects(trailhead_buffer)]
    
    # Remove those from both datasets
    ps_mod = gdf_parkserve[~gdf_parkserve.index.isin(ps_near_th.index)]
    osm_mod = gdf_osm[~gdf_osm.index.isin(osm_near_th.index)]
    
    return ps_mod, osm_mod, trailhead_polys

```

```python
ps_mod, osm_mod, trail_poly = get_trailhead_polygons(
    place="Spokane County, Washington",
    gdf_parkserve=gdf,
    gdf_osm=gdf_osm,
    area_threshold=200000,
    grid_size=150
)
print("Done!")


# Convert to lat/lon
ps_mod_latlon = ps_mod.to_crs(4326)
osm_mod_latlon = osm_mod.to_crs(4326)
trail_poly_latlon = trail_poly.to_crs(4326)

center = ps_mod_latlon.unary_union.centroid.coords[0][::-1]
m = folium.Map(location=center, zoom_start=10, tiles="cartodbpositron")

# ParkServe
folium.GeoJson(
    ps_mod_latlon,
    name="ParkServe (filtered)",
    style_function=lambda x: {"color": "green", "weight": 1, "fillOpacity": 0.2}
).add_to(m)

# OSM
folium.GeoJson(
    osm_mod_latlon,
    name="OSM Parks (filtered)",
    style_function=lambda x: {"color": "gray", "weight": 1, "fillOpacity": 0.2}
).add_to(m)

# Trailhead polygons (orange squares)
folium.GeoJson(
    trail_poly_latlon,
    name="Trailhead Polygons",
    style_function=lambda x: {"color": "red", "fillColor": "red", "weight": 5, "fillOpacity": 0.6}
).add_to(m)

folium.LayerControl().add_to(m)
m.save("trailhead_parks_filtered.html")
m
```
